# โจทย์ Bayesian Network: Smoker → Cancer → X-Ray


## 1. แปลงตารางเป็นตัวเลข

```python
P_S      = 0.30  
P_C_St   = 0.05   
P_C_Sf   = 0.01  
P_T_Ct   = 0.90  
P_T_Cf   = 0.20 
```

## ข้อ 41 — `P(S=true, C=true, T=บวก)`

```python
ans1 = P_S * P_C_St * P_T_Ct
print(f"P(S=t, C=t, T=บวก) = {P_S} × {P_C_St} × {P_T_Ct} = {ans1}")
print(f"→ ตอบข้อ b) 0.0135")
```

## ข้อ 42 — `P(T=บวก)`

```python
P_C = P_C_St * P_S + P_C_Sf * (1 - P_S)
print(f"P(C=true) = {P_C_St}×{P_S} + {P_C_Sf}×{1-P_S:.1f} = {P_C:.4f}  มะเร็งพบเพียง {P_C*100:.1f}%")

P_T = P_T_Ct * P_C + P_T_Cf * (1 - P_C)
print(f"P(T=บวก)  = {P_T_Ct}×{P_C:.3f} + {P_T_Cf}×{1-P_C:.3f} = {P_T:.4f}")
print(f"ตอบข้อ b) 0.2154")
```

## ข้อ 43 — `P(C=true | T=บวก)`  

```python
joint_C_T = P_T_Ct * P_C         
ans43 = joint_C_T / P_T
print(f"P(C=t, T=บวก) = {P_T_Ct} × {P_C:.4f} = {joint_C_T:.4f}")
print(f"P(C=t | T=บวก) = {joint_C_T:.4f} / {P_T:.4f} = {ans43:.4f}")
print(f"ตอบข้อ b) 0.092")
 
```

## ข้อ 44 — `P(S=true | T=บวก)`  (Posterior ของการสูบบุหรี่)

```python
true_pos  = P_C_St * P_T_Ct       
false_pos = (1 - P_C_St) * P_T_Cf 
joint_S_T = P_S * (true_pos + false_pos)

print(f"เส้นทางเป็นมะเร็งจริง : {P_C_St} × {P_T_Ct} = {true_pos:.4f}")
print(f"เส้นทางผลบวกลวง      : {1-P_C_St} × {P_T_Cf} = {false_pos:.4f}")
print(f"P(S=t, T=บวก) = {P_S} × ({true_pos:.3f} + {false_pos:.3f}) = {joint_S_T:.4f}")

ans44 = joint_S_T / P_T
print(f"P(S=t | T=บวก) = {joint_S_T:.4f} / {P_T:.4f} = {ans44:.4f}")
print(f"ตอบข้อ a) 0.327")
 
```

## ข้อ 45 — ผู้ป่วย 1,000 คน เป็นมะเร็ง **และ** ผลบวก กี่ราย?

```python
N = 1000
ans45 = N * joint_C_T
print(f"{N} × {joint_C_T:.4f} = {ans45:.1f} ราย")
print(f"ตอบข้อ b) ≈ 20")
```
